## Choose an investigation route

All routes use the same marking criteria.

- **A — Investigate the existing architecture:** extend or deepen one of the practical investigations.
- **B — Make an important change:** replace or substantially modify one component and test the consequences.
- **C — Build an alternative architecture:** for example, compare a Transformer + CLIP approach with the reference architecture.
- **D — Ask a scientific question:** test a claim about what or how the architecture learns, for example whether it learns a concept of place.
